# NETFLIX EDA PROJECT

### IMPORT LIBRARY

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Read Netflix Data

```python
df = pd.read_csv("/content/NetFlix.csv")
df
```

### Total Columns in Netflix Data

```python
df.columns
```

### Cleaning Data Remove Description

```python
df.drop(['description'], axis=1, inplace=True)
```

### Q1: What is the count of Movies vs TV Shows on Netflix?
####Using this chart: Bar Chart

```python
df['type'].value_counts().plot(kind='bar', color=['teal', 'coral'])
plt.title('Count of Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()

df['type'].value_counts()
```

### Q2: What is the frequency distribution of Movie durations?
####Using this chart: Histogram

```python
df[df['type'] == 'Movie']['duration'].plot(kind='hist', bins=20, color='red')
plt.title('Movie Duration Distribution')
plt.xlabel('Minutes')
plt.ylabel('Frequency')
plt.show()
```

### Q3: Which 5 countries have the highest amount of content on Netflix?
####Using this chart: Pie Chart

```python
top5_countries = df['country'].dropna().str.split(', ').explode().value_counts().head(5)

top5_countries.plot(kind='pie', autopct='%1.1f%%', startangle=140)
plt.title('Top 5 Content Producing Countries')
plt.ylabel('')
plt.show()

top5_countries
```

### Q4: How has the number of releases grown from year 2000 to 2021?
####Using this chart: Line Chart

```python
df[df['release_year'] >= 2000]['release_year'].value_counts().sort_index().plot(kind='line', marker='o')
plt.title('Releases Per Year (2000-2021)')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.show()
```

### Q5: What are the top 5 most common content ratings on Netflix?
####Using this chart: Donut Chart

```python
top_ratings = df['rating'].value_counts().head(5)

top_ratings.plot(kind='pie', autopct='%1.1f%%', startangle=90, wedgeprops=dict(width=0.5))
plt.title('Top 5 Most Common Content Ratings')
plt.ylabel('')
plt.show()

top_ratings
```

### Q6: How has the release volume of Movies vs TV Shows evolved from 2010 to 2021?
####Using this chart: Heatmap

```python
recent_df = df[df['release_year'] >= 2010]
heat_data = pd.crosstab(recent_df['release_year'], recent_df['type'])

sns.heatmap(heat_data, annot=True, fmt='d', cmap='YlGnBu', cbar=False)
plt.title('Movie vs TV Show Releases (2010–2021)')
plt.show()
```

# Netflix EDA Project

A Exploratory Data Analysis (EDA) project on the Netflix dataset to analyze movies, TV shows, release trends, content ratings, and top producing countries using Python.

---

## 📌 Project Overview
This project explores Netflix catalog data to understand:
- The split between Movies and TV Shows.
- Runtime distribution of movies.
- Top content-producing countries.
- Release trends from 2000 to 2021.
- Yearly release patterns using heatmaps.
- Most common content ratings.

---

## 🛠️ Tech Stack
- **Language:** Python
- **Environment:** Google Colab / Jupyter Notebook
- **Libraries:** Pandas, NumPy, Matplotlib, Seaborn

---

## 📊 Key Insights & Visualizations

- Q1: Movies vs. TV Shows (Bar Chart):

  - Movies make up ~69% (5,377) and TV Shows make up ~31% (2,410) of the catalog.
- Q2: Movie Duration (Histogram):

  - Most movies have a duration between 80 and 110 minutes.
- Q3: Content by Country (Pie Chart):

  - The United States leads content production, followed by India and the United Kingdom.
- Q4: Growth Over Years (Line Chart):

  - Content production grew rapidly after 2014 and peaked in 2018.
- Q5: Release Trend by Type (Heatmap):

  - Movie releases peaked in 2017–2018, while TV shows peaked in 2020.
- Q6: Content Ratings (Donut Chart):

  - TV-MA and TV-14 are the most common audience ratings.
